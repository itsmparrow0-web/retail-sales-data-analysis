```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print ("All libraries imported sucessfully!")
```

```python

data = {
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115],

    "City": ["Bangalore", "Delhi", "Mumbai", "Bangalore", "Delhi", "Mumbai", "Bangalore", "Delhi", "Mumbai", "Bangalore", "Delhi", "Mumbai", "Bangalore", "Delhi", "Mumbai"],

    "Product": ["Laptop", "Mobile", "Laptop", "Headphones", "Mobile", "Tablet", "Laptop", "Headphones", "Mobile", "Tablet", "Laptop", "Headphones", "Mobile", "Tablet", "Laptop"],

    "Quantity": [2, 5, 1, 10, 3, 4, 3, 8, 6, 2, 2, 12, 4, 3, 2],

    "Price": [45000, 15000, 45000, 1000, 15000, 20000, 45000, 1000, 15000, 20000, 45000, 1000, 15000, 20000, 45000],

    "Month": ["January", "January", "January", "February", "February", "February", "March", "March", "March", "March", "April", "April", "April", "April", "May"]
}

df = pd.DataFrame(data)

print("Company Sales Dataset Created Successfully!")
```

```python

df["Revenue"] = df["Quantity"] * df["Price"]

print("Revenue calculated successfully!")

df[["OrderID", "Product", "Quantity", "Price", "Revenue"]]
```

```python
total_revenue = df["Revenue"].sum()

print("Total Company Revenue:", total_revenue)
```

```python

product_sales = df.groupby("Product")["Quantity"].sum()

print(product_sales.sort_values(ascending=False))
```

```python

# Calculate city-wise revenue

city_revenue = df.groupby("City")["Revenue"].sum()

print("City-wise Revenue:")
print(city_revenue.sort_values(ascending=False))

# Create Bar Graph

plt.figure(figsize=(8, 5))

city_revenue.sort_values(ascending=False).plot(kind="bar")

plt.title("City-wise Revenue Analysis")
plt.xlabel("City")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

```python

# Monthly Revenue Analysis

monthly_revenue = df.groupby("Month")["Revenue"].sum()

month_order = ["January", "February", "March", "April", "May"]

monthly_revenue = monthly_revenue.reindex(month_order)

print("Monthly Revenue:")
print(monthly_revenue)

# Create Line Graph

plt.figure(figsize=(9, 5))

plt.plot(monthly_revenue.index, monthly_revenue.values,
         marker="o", linewidth=2)

plt.title("Monthly Sales Revenue Analysis")
plt.xlabel("Month")
plt.ylabel("Revenue (INR)")
plt.grid(True)

plt.show()
```
